# Getting Started with TensorFlow


```python
import tensorflow as tf
import numpy as np

# set seeds for reproduciblity

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)
```

```python
scalar = tf.constant(7)
vector = tf.constant([[10, 10],[10, 10]])
matrix = tf.constant([[10, 7],
                      [7, 10]])

tensor_3d = tf.random.uniform(shape=(2, 3, 4), minval=0, maxval=10, dtype=tf.int32)

tensor_4d = tf.random.uniform(shape=(2, 3, 4, 5), minval=0, maxval=10, dtype=tf.int32)

print(f"Shape {scalar.shape} ndim = {scalar.ndim}")
print(f"Shape {vector.shape} ndim = {vector.ndim}")
print(f"Shape {matrix.shape} ndim = {matrix.ndim}")
print(f"Shape {tensor_3d.shape} ndim = {tensor_3d.ndim}")
print(f"Shape {tensor_4d.shape} ndim = {tensor_4d.ndim}")
```

## Creating Tensors

```python
from_list = tf.constant(
    [
        [1,2,3,4,5],
        [32,324,554,4554,34]
    ])



from_numpy = tf.constant(np.array([[1,2,3,4,5],[32,324,554,4554,34]]))


ones = tf.ones(shape=(3,4))
zeros = tf.zeros(shape=(3,4))

rand = tf.random.uniform(shape=(3,4),maxval=1, minval=-1)
randn = tf.random.normal(shape=(3,4))
arrange = tf.range(start=1, limit=10, delta=2)

print(from_numpy)
print(from_list)

print("*********************************")

print(f" Range = {arrange}")
print(f"normal = {randn}")
print(f"uniform = {rand}")
print(f"ones = {ones}")
print(f"zeros = {zeros}")

print("*********************************")

print(ones)
print(zeros)
```

```python
c = tf.constant([1,2,3])
v = tf.Variable([1,2,3])

print(c,v)
```
